### Refuerzo
 Escribe tu propio ticket como diccionario: una clave "tienda" y una clave "lineas" con una lista de dos productos, cada uno con su nombre, sus unidades y su precio. Calcula el total accediendo a cada dato por su clave y su posición.

```python
ticket = {
    "tienda": "GranVía",
    "lineas": [
        {"nombre": "Pan de molde integral", "unidades": 2, "precio": 1.45},
        {"nombre": "Aceite de oliva 1 L", "unidades": 1, "precio": 4.20}
    ]
}

producto1 = ticket["lineas"][0]["nombre"]
unidades1 = ticket["lineas"][0]["unidades"]
precio1 = ticket["lineas"][0]["precio"]

producto2 = ticket["lineas"][1]["nombre"]
unidades2 = ticket["lineas"][1]["unidades"]
precio2 = ticket["lineas"][1]["precio"]

total = (unidades1 * precio1) + (unidades2 * precio2)

print(f"Tienda: {ticket['tienda']}")
print(f"Línea 1: {producto1} x{unidades1} = {unidades1 * precio1:.2f}€")
print(f"Línea 2: {producto2} x{unidades2} = {unidades2 * precio2:.2f}€")
print(f"Total: {total:.2f}€")
```

### Ampliación

 El total del pedido PED0170755 es 5,96 €. Comprueba con Python que coincide con la suma de sus tres líneas. Cuidado: la primera lleva un 5 % de descuento. El resultado tiene que ser 5.96.

```python
pedido = {
    "pedido_id": "PED0170755",
    "fecha": "2025-01-15T20:20:27",
    "cliente": {"id": "C01384", "provincia": "Madrid", "tiene_app": True},
    "tienda_id": "T006",
    "lineas": [
        {"producto_id": "P0165", "nombre": "Zumo de piña 1 L Colmena", "categoria": "Bebidas",
         "unidades": 1, "precio": 1.11, "descuento_pct": 5},
        {"producto_id": "P0101", "nombre": "Kéfir 500 ml Colmena", "categoria": "Lácteos y huevos",
         "unidades": 1, "precio": 1.38},
        {"producto_id": "P0268", "nombre": "Cacao soluble 500 g El Molinillo", "categoria": "Despensa",
         "unidades": 1, "precio": 3.53},
    ],
    "total": 5.96,
    "entrega": {"tipo": "domicilio", "cp": "28037", "franja": "tarde", "coste_envio": 4.95},
    "pago": {"metodo": "tarjeta"},
    "estado": "entregado",
    "valoracion": 3,
}
```

```python
linea1 = pedido["lineas"][0]
linea2 = pedido["lineas"][1]
linea3 = pedido["lineas"][2]

importe1 = linea1["unidades"] * linea1["precio"] * (1 - linea1["descuento_pct"] / 100)
importe2 = linea2["unidades"] * linea2["precio"]
importe3 = linea3["unidades"] * linea3["precio"]

suma_total = importe1 + importe2 + importe3

print(f"Suma de las líneas: {suma_total:.2f}€")
print(f"Total del pedido: {pedido['total']}€")
print(f"¿Coinciden?: {round(suma_total, 2) == pedido['total']}")
```
