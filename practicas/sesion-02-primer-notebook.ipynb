### Tarea 3 · Las cámaras de frío, con variables

Cada tienda tiene 3 cámaras de frío, y cada cámara envía su temperatura 12 veces por hora, las 24 horas del día.

Crea una variable para cada dato: tiendas, camaras_por_tienda y lecturas_por_hora.
Calcula lecturas_al_dia a partir de ellas.
Calcula lecturas_al_anio. Los sensores no descansan: son 365 días.
Muestra los dos resultados con print y una f-string.

```python
tiendas = 120
camaras_por_tienda = 3
lecturas_por_hora = 12
horas_por_dia = 24
dias_por_anio = 365


lecturas_al_dia = tiendas * camaras_por_tienda * lecturas_por_hora * horas_por_dia
lecturas_al_anio = lecturas_al_dia * dias_por_anio

print(f"Lecturas: {lecturas_al_dia:,} al día")
print(f"Lecturas: {lecturas_al_anio:,} al año")
```

### Tarea 4 · De lecturas a megas
Cada lectura ocupa unos 46 bytes. ¿Cuánto ocupan las lecturas de un día? ¿Y las de un año?

Crea la variable bytes_por_lectura.
Calcula megas_al_dia. Recuerda que 1 MB son 1.000.000 de bytes.
Calcula gigas_al_anio. Recuerda que 1 GB son 1.000 MB.
Muestra los resultados con dos decimales, usando :.2f dentro de las llaves.

```python
bytes_por_lectura = 46
bytes_por_MB = 1_000_000
bytes_por_GB = 1_000_000_000

megas_al_dia = (lecturas_al_dia * bytes_por_lectura) / bytes_por_MB
gigas_al_anio = (lecturas_al_anio * bytes_por_lectura) / bytes_por_GB

print(f"Sensores: {megas_al_dia:.2f} MB al día")
print(f"Sensores: {gigas_al_anio:.2f} GB al año")
```

### Tarea 5 · Los clics de la web y la app
